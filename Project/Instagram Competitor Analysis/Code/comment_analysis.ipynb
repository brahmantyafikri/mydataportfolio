```python
import re

def clean_text(s):
    s = str(s).lower()
    s = re.sub(r"http\S+|www\.\S+", " ", s)
    s = re.sub(r"@\w+", " ", s)
    s = re.sub(r"#\w+", " ", s)
    s = re.sub(r"[^a-z0-9\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

df["text_clean"] = df["comment_text"].apply(clean_text)
df = df[df["text_clean"].str.len() > 2].copy()
```

```python
INTENT_KEYWORDS = {
    "respondent_interest": ["responden", "survey", "survei", "kuesioner", "isi", "link"],
    "pricing": ["harga", "biaya", "price", "tarif", "paket"],
    "complaint": ["bohong", "penipuan", "scam", "hoax", "gagal", "error", "salah", "ripoff"],
    "trust_legitimacy": ["akurat", "ilmiah", "valid", "percaya", "terpercaya", "kredibel"],
    "career": ["lowongan", "magang", "intern", "kerja", "rekrut"],
    "partnership": ["kerja sama", "kolaborasi", "partner", "partnership"],
}

priority = ["complaint","pricing","respondent_interest","partnership","career","trust_legitimacy"]

def detect_intent(t):
    for intent in priority:
        if any(k in t for k in INTENT_KEYWORDS[intent]):
            return intent
    return "other"

df["intent"] = df["text_clean"].apply(detect_intent)
```

```python
POS = ["bagus","mantap","keren","nice","good","hebat","sukses","terima kasih","amazing","wow"]
NEG = ["jelek","buruk","bohong","penipu","penipuan","scam","hoax","salah","gagal","parah"]

def lex_sentiment(t):
    pos = sum(w in t for w in POS)
    neg = sum(w in t for w in NEG)
    if pos > neg and pos > 0: return "positive"
    if neg > pos and neg > 0: return "negative"
    return "neutral"

df["sentiment"] = df["text_clean"].apply(lex_sentiment)
```

```python
import pandas as pd

def top_posts(sub, n=3):
    return (sub["post_url"].value_counts().head(n)).to_dict()

summary_rows = []
for owner, sub in df.groupby("ownerUsername"):
    summary_rows.append({
        "ownerUsername": owner,
        "n_comments": len(sub),
        "n_posts": sub["post_url"].nunique(),
        "top_intent": sub["intent"].value_counts().idxmax(),
        "top_sentiment": sub["sentiment"].value_counts().idxmax(),
        "top_posts": top_posts(sub, n=3),
    })

summary_df = pd.DataFrame(summary_rows).sort_values("n_comments", ascending=False)
summary_df
```

```python
def show_owner_report(owner, n_examples=5):
    sub = df[df["ownerUsername"] == owner].copy()
    print("="*70)
    print(f"OWNER: @{owner}")
    print("Komentar:", len(sub), "| Post unik:", sub["post_url"].nunique())

    print("\nIntent distribution:")
    print(sub["intent"].value_counts())

    print("\nSentiment distribution:")
    print(sub["sentiment"].value_counts())

    print("\nTop commented posts:")
    print(sub["post_url"].value_counts().head(5))

    print("\nContoh komentar (berdasarkan intent dominan):")
    top_intent = sub["intent"].value_counts().idxmax()
    ex = sub[sub["intent"] == top_intent][["comment_text","post_url"]].head(n_examples)
    display(ex)

# contoh panggil
for owner in df["ownerUsername"].unique():
    show_owner_report(owner, n_examples=5)
```

```python
from collections import Counter

STOPWORDS_ID = set(["yang","dan","di","ke","dari","ini","itu","untuk","nya","aja","kok","kak","bang","bro","the","a","an","to","of","in","on","you","me","can"])

def top_words(text_series, n=15):
    words = []
    for t in text_series:
        words.extend([w for w in t.split() if w not in STOPWORDS_ID and len(w) > 2])
    return Counter(words).most_common(n)

for owner, sub in df.groupby("ownerUsername"):
    print("\n", "="*20, owner, "="*20)
    print(top_words(sub["text_clean"], n=15))

```
